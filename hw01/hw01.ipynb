````python
# 10
unique_cities = list()
revenue_by_city = list()
city_and_revenue = list()
for c, a, p in zip(cities_clean, amounts_int, paid): #сортировка какие платежи прошли
    if p:
        city_and_revenue.append(list((c, a)))

revenue = list() #подготовка к фильтрации по прибыли
for i in city_and_revenue:
    revenue.append(i[1])
sorted_revenue = sorted(revenue, reverse=True)

for i in sorted_revenue: #сортировка по прибыли 
    for j in city_and_revenue:
        if i == j[1]:
            revenue_by_city.append(j)

for i in revenue_by_city: #сортировка городов 
    if i[0] not in unique_cities:
        unique_cities.append(i[0])
print(revenue_by_city)
print(unique_cities)
````

````python
#12
dayOfTheWeekend = []
for i in days:
    if i not in dayOfTheWeekend:
        dayOfTheWeekend.append(i)

payments = {}
for a, p, d in zip(amounts_int, paid, days):
    if p:
        payments[d] = payments.get(d, 0) + a

for i in payments:
    print(f"{i:>5} | {"█" * (payments[i] // 5000):<8}  {payments[i]:>7}")
````
