````python
#7
import time 
amounts = [float(r["amount"].replace(" ", "")) for r in rows]
paids = [r["status"] == "оплачен" for r in rows]

t = time.time()
total = 0
for a, p in zip(amounts, paids):
    if p:
        total += a
t_loop = time.time() - t

t = time.time()
amount[paid].sum()
t_np = time.time() - t
print(f"Цикл: {t_loop:.5f} c, NumPY: {t_np:.5f} c, быстрее в {t_loop / t_np:.0f} раз")
````

````python
#8
np.random.seed(42)
a, b = 0, 0
for i in range(100):
    group_a = (np.random.rand(1000) < 0.05).sum()
    group_b = (np.random.rand(1000) < 0.06).sum()
    print(f"Количество покупок в группе A: {group_a} | в группе B: {group_b}")
    if group_a > group_b:
        a += 1
    else:
        b += 1
print(f"Группа B оказалась лучше группы A на {b / (a+b) * 100} %")
````

Исходя из анализа,  мы можем наблюдать что из 100 попыток группа B побеждает группу А в 85% случаев. Увеличение размера выборки (количества человек) уменьшает влияние случайности, тем самым мы уменьшаем количество шума в данных, что даёт нам более правдивые результаты
