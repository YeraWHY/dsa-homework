## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База
````

````python
#1
def to_num(s):
    return float(s.replace(" ", "")) 
print(to_num(rows[4954]["amount"])) #проверка


cities = {c["city"].strip().capitalize() for c in rows} #хотел посмотреть встречаются ли вообще в данных города на латинице и грязные ли они

def clean_city(s):
    cities = {'Almaty': 'Алматы',
             'Astana': 'Астана',
              'Shymkent': 'Шымкент',
              'Aktobe': 'Актобе',
              'Atyrau': 'Атырау',
              'Karaganda': 'Караганда'
             }
    if s.strip().capitalize() in cities:
        return cities[s.strip().capitalize()]
    else:
        return s.strip().capitalize()

print(clean_city("  almaty  ")) #проверка
````

````python
#2
order_category = {}
for i in rows:
    order_category[i["category"]] = order_category.get(i["category"], 0) + 1
order_category
````

````python
#3 
paid_category = {}
for i in rows:
    if i["status"] == "оплачен":
        paid_category.setdefault(i["category"], []).append(to_num(i["amount"]))
for i in paid_category:
    print(f"{i:<14} Выручка:{sum(paid_category[i]):<10} ₸ | Средний чек:{round(sum(paid_category[i]) / len(paid_category[i]))}")
````

````python
#4
def clear_month(m):
    m = list(m.replace("-", "").replace(" ", "").replace(":", ""))
    return "".join(m[:4]) + "-" + "".join(m[4:6])

total_in_month_order = {}
for i in rows:
    total_in_month_order[clear_month(i["order_date"])] = total_in_month_order.get(clear_month(i["order_date"]), 0) + 1
total_in_month_order
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
````

````python
#5
paid_in_month_order = {}
for i in rows:
    if i["status"] == 'оплачен':
        paid_in_month_order.setdefault(clear_month(i["order_date"]), []).append(to_num(i["amount"]))
for j in paid_in_month_order:
    print(f"{j} Выручка:{sum(paid_in_month_order[j]):<10}")
print(f"Изменения выручки:{(sum(paid_in_month_order["2026-10"]) - sum(paid_in_month_order["2026-09"])) / sum(paid_in_month_order["2026-09"]) * 100:.2f} %")
````

````python
#6
paid_orders = {}
for i in rows:
    if i["status"] == "оплачен":
        paid_orders[clean_city(i["city"])] = paid_orders.get(i["city"], 0) + 1
paid_orders
unpaid_orders = {}
for i in rows:
    if i["status"] == "отменён":
        unpaid_orders[clean_city(i["city"])] = unpaid_orders.get(i["city"], 0) + 1

for j in sorted(paid_orders):
    print(f"{j:<10}Разница между статусами:{unpaid_orders[j] / (paid_orders[j] + unpaid_orders[j]) * 100:.2f} %")
````

````python
#7
paid_customer = {}
for i in rows:
    if i["status"] == "оплачен":
        paid_customer.setdefault(i["customer_id"], []).append(to_num(i["amount"]))

for i, j in list(sorted(paid_customer.items(), key=lambda t: sum(t[1]), reverse=True))[:5]:
    print(i, j)
````

На первом месте клиент с id - "TEST" с ОЧЕНЬ большими затратами. На основе его трат могу сделать вывод что это тестовые данные для проверки навыков на собеседовании.  

````python
#8
unique_clients = {}
for i in rows:
    unique_clients.setdefault(clean_city(i["city"]), set()).add(i["customer_id"])
for i in unique_clients:
    print(f"В городе {i:<12} {len(unique_clients[i]):>5} уникальных клиентов")
````

````python
# Челлендж
````

````python
#9
new_rows = set()
dublicates_in_month = {}
for i in rows:
    if tuple(i.items()) not in new_rows:
        new_rows.add(tuple(i.items()))
    else:
        dublicates_in_month[clear_month(i["order_date"])] = dublicates_in_month.get(clear_month(i["order_date"]), 0) + 1
print(f"Количество не дубликатов:{len(new_rows):<5} | Сколько дубликатов и в каком мясеце:{dublicates_in_month}")
````

````python
#10 
total_in_month = {}
for i in new_rows:
    row_dict = dict(i)
    if row_dict["status"] == 'оплачен' and row_dict["customer_id"] != "TEST":
        total_in_month.setdefault(clear_month(row_dict["order_date"]), []).append(to_num(row_dict["amount"]))
for j in total_in_month:
    print(f"{j} Выручка:{sum(total_in_month[j]):<10}")
print(f"Изменения выручки:{(sum(total_in_month["2026-10"]) - sum(total_in_month["2026-09"])) / sum(total_in_month["2026-09"]) * 100:.2f} %")

#разница почти в 20 % после уборки дубликатов и тестовых данных
````

````python
#11 
more_orders = {}
for i in new_rows:
    row_dict = dict(i)
    more_orders[row_dict["customer_id"]] = more_orders.get(row_dict["customer_id"], 0) + 1

best = max(more_orders, key=more_orders.get)
print(f"Больше всего заказов у {best} = {more_orders[best]}")
````
